```python
import matplotlib.pyplot as plt
import numpy as np

plt.suptitle("GRAPH")
x = np.array([0,4])
y = np.array([3,8])

plt.subplot(2, 3, 1)
plt.title("title1")
plt.xlabel("x label",color="blue")
plt.ylabel("y label",color="blue")


plt.plot(x,y,"o")



x = np.array([10,40,60, 80,100])
y = np.array([60, 20, 70, 20,20])

w = np.array([23,43,60, 78,100])
z = np.array([6, 40, 8, 90,90])

plt.subplot(2, 3, 2)
plt.plot(x,y,linestyle="dashed")
plt.plot(w,z,marker="*",linewidth=2.5)



x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2, 3, 3)


plt.plot(x,y,marker="o",mfc="yellow",mec="black")

x = np.array([0, 100])
y = np.array([0, 100])

plt.subplot(2, 3, 4)
plt.title("title2")
plt.grid(axis='x')
plt.plot(x,y)

x = np.array([0, 1, 2, 3,5,3,7,4,8,9,3,5,7])
y = np.array([3, 8, 1, 10,3,7,9,2,0,4,5,6,7])
colors = np.array(["red","green","blue","yellow","pink","black","orange","purple","beige","brown","gray","cyan","magenta"])
plt.subplot(2, 3, 5)
sizes = np.array([22, 455, 333, 22,33,23,7,32,55,3,44,5,600])
plt.scatter(x, y, s=sizes,c = colors, alpha=0.5)

x=np.array(["A","B","C"])
y=np.array([6,3,8])


plt.subplot(2, 3, 6)
plt.bar(x,y,color ="r",width=0.1)

plt.show()

```
